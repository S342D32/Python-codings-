Create a Node and linkedlist

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.n = 0

    def __len__(self):
        return self.n

    def insert_head(self, value):
        # New node
        new_node = Node(value)
        # Create connection
        new_node.next = self.head
        # Reassign head
        self.head = new_node
        # Increment count
        self.n += 1

    def __str__(self):
        curr = self.head
        result = ''
        while curr is not None:
            result += str(curr.value) + '->'
            curr = curr.next
        # Remove the last arrow in the output
        return result[:-2] if result else 'Empty list'

    def insert_middle(self, after, value):
        new_node = Node(value)
        curr = self.head
        while curr is not None:
            if curr.value == after:
                # Found the node after which we want to insert
                break
            curr = curr.next
        if curr is not None:
            # Insert the new node in the middle
            new_node.next = curr.next
            curr.next = new_node
            self.n += 1
        else:
            return 'Item not found'

    def clear(self):
        # Clear the entire linked list
        self.head = None
        self.n = 0

    def delete(self):
        if self.head is None:
            return 'Empty list'
        else:
            # Delete the head node
            self.head = self.head.next
            self.n -= 1

    def pop(self):
        if self.head is None:
            return 'Empty list'
        elif self.head.next is None:
            # If only one element in the list
            self.head = None
        else:
            # Traverse to the second-last node
            curr = self.head
            prev = None
            while curr.next is not None:
                prev = curr
                curr = curr.next
            # Remove the last node
            prev.next = None
        self.n -= 1

    def search(self, item):
        curr = self.head
        pos = 0
        while curr is not None:
            if curr.value == item:
                return pos
            curr = curr.next
            pos += 1
        return 'Item not found'

    # Magic method for indexing
    def __getitem__(self, index):
        curr = self.head
        pos = 0
        while curr is not None:
            if pos == index:
                return curr.value
            curr = curr.next
            pos += 1
        return 'Index Error'

    def append(self, value):
        new_node = Node(value)
        if self.head is None:
            # If the list is empty, set head to the new node
            self.head = new_node
        else:
            curr = self.head
            while curr.next is not None:
                curr = curr.next
            # We are at the last node, append the new node
            curr.next = new_node
        self.n += 1

    def sum_odd_nodes(self):
        temp = self.head
        count = 0
        result = 0
        while temp is not None:
            if count % 2 != 0:
                result += temp.value
            count += 1
            temp = temp.next
        print(result)
   
    


    def traverse(self):
        curr = self.head
        while curr is not None:
            print(curr.value, end=' ')
            curr = curr.next
        print()
    
    


```

Reversing LinkedList

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.n = 0

    def __len__(self):
        return self.n


    def __str__(self):
        curr = self.head
        result = ''
        while curr is not None:
            result += str(curr.value) + '->'
            curr = curr.next
        return result[:-2] if result else 'Empty list'

    

    def append(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
        else:
            curr = self.head
            while curr.next is not None:
                curr = curr.next
            curr.next = new_node
        self.n += 1
        
    def traverse(self):
        curr = self.head
        while curr is not None:
            print(curr.value, end=' ')
            curr = curr.next
        print()

    def reverse(self):
        prev = None
        curr = self.head
        while curr is not None:
            next_node = curr.next
            curr.next = prev
            prev = curr
            curr = next_node
        self.head = prev

# Example usage
ll = LinkedList()
ll.append(1)
ll.append(2)
ll.append(3)
ll.append(4)

print("Linked List:")
ll.traverse()  # Output: 1 2 3 4

ll.reverse()
print("Reversed Linked List:")
ll.traverse()  # Output: 4 3 2 1
```
